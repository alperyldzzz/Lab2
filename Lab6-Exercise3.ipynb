```python
import copy
import numpy as np
import scipy.linalg


def linearsolver(A, b):
    n = len(A)

    # Augmented matrix [A|b]
    M = []
    for i in range(n):
        row = A[i] + [b[i]]
        M.append(row)

    # Gaussian elimination with partial pivoting
    for k in range(n):
        # Pivoting
        for i in range(k, n):
            if abs(M[i][k]) > abs(M[k][k]):
                M[k], M[i] = M[i], M[k]

        # Elimination
        for j in range(k + 1, n):
            q = float(M[j][k]) / M[k][k]
            for m in range(k, n + 1):
                M[j][m] -= q * M[k][m]

    # Back substitution
    x = [0 for _ in range(n)]

    x[n - 1] = float(M[n - 1][n]) / M[n - 1][n - 1]

    for i in range(n - 1, -1, -1):
        z = 0
        for j in range(i + 1, n):
            z += float(M[i][j]) * x[j]
        x[i] = float(M[i][n] - z) / M[i][i]

    return x


A = [
    [4, -1, 1],
    [2, 5, 2],
    [1, 2, 4]
]

b = [8, 3, 11]

A_copy = copy.deepcopy(A)
b_copy = b.copy()

x_gauss = linearsolver(A_copy, b_copy)

A_np = np.array(A, dtype=float)
b_np = np.array(b, dtype=float)

x_scipy = scipy.linalg.solve(A_np, b_np)

print("Exercise 3 - Gaussian Elimination with Pivoting")
print("----------------------------------------------")

print("Solution using Gaussian elimination with pivoting:")
print(x_gauss)

print("\nSolution using scipy.linalg.solve:")
print(x_scipy)
```
