```python
import numpy as np
import scipy.integrate as integrate


def f(x):
    return x**2


a = 1
b = 4
n = 3
h = (b - a) / n

exact_value, _ = integrate.quad(f, a, b)

x_values = np.array([1, 2, 3, 4], dtype=float)

left_points = x_values[:-1]
left_result = h * np.sum(f(left_points))

right_points = x_values[1:]
right_result = h * np.sum(f(right_points))

mid_points = np.array([1.5, 2.5, 3.5])
mid_result = h * np.sum(f(mid_points))

trap_result = h * ((f(a) + f(b)) / 2 + np.sum(f(x_values[1:-1])))

print("Lab 7 - Exercise 1 and Exercise 2")
print("--------------------------------")
print("Integral: from 1 to 4 of x^2 dx")
print("Exact analytical value:", exact_value)
print()

print("Left rectangle result:", left_result)
print("Left rectangle error:", abs(exact_value - left_result))
print()

print("Right rectangle result:", right_result)
print("Right rectangle error:", abs(exact_value - right_result))
print()

print("Midpoint rectangle result:", mid_result)
print("Midpoint rectangle error:", abs(exact_value - mid_result))
print()

print("Trapezoidal result:", trap_result)
print("Trapezoidal error:", abs(exact_value - trap_result))
```
