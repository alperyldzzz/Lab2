```python
import numpy as np
import matplotlib.pyplot as plt


def f(x):
    return x * np.sin(4 * x)


def df(x):
    return np.sin(4 * x) + 4 * x * np.cos(4 * x)


def newton_method(x0, tol_x, tol_f, max_iter):
    x_old = x0

    for iteration in range(1, max_iter + 1):
        derivative = df(x_old)

        if derivative == 0:
            return None, iteration, False

        x_new = x_old - f(x_old) / derivative

        if abs(x_new - x_old) <= tol_x:
            return x_new, iteration, True

        if abs(f(x_new)) <= tol_f:
            return x_new, iteration, True

        x_old = x_new

    return x_new, max_iter, False


tol_x = 1e-6
tol_f = 1e-6
max_iter = 100


print("PART 1: Newton Method for One Initial Guess")
print("------------------------------------------")

example_guess = 1.0
root, iteration_count, converged = newton_method(example_guess, tol_x, tol_f, max_iter)

print("Initial guess:", example_guess)
print("Found root:", root)
print("Number of iterations:", iteration_count)
print("Converged:", converged)


print("\nPART 2: Newton Method for Different Initial Guesses")
print("--------------------------------------------------")

initial_guesses = np.arange(-5, 5, 0.05)

roots = []
iterations = []
convergence = []

for x0 in initial_guesses:
    root, iteration_count, converged = newton_method(x0, tol_x, tol_f, max_iter)

    roots.append(root)
    iterations.append(iteration_count)
    convergence.append(converged)


print("Initial Guess     Found Root        Iterations     Converged")
print("------------------------------------------------------------")

for i in range(len(initial_guesses)):
    if roots[i] is None:
        print(f"{initial_guesses[i]:10.4f}        None          {iterations[i]:5d}        {convergence[i]}")
    else:
        print(f"{initial_guesses[i]:10.4f}     {roots[i]:10.6f}        {iterations[i]:5d}        {convergence[i]}")


plt.figure(figsize=(10, 6))
plt.plot(initial_guesses, roots, '.', markersize=5)
plt.xlabel("Initial guess")
plt.ylabel("Found root")
plt.title("Newton Method: Initial Guess vs Found Root")
plt.grid(True)
plt.savefig("initial_guess_vs_found_root.png")
plt.show()


plt.figure(figsize=(10, 6))
plt.plot(initial_guesses, iterations, '.', markersize=5)
plt.xlabel("Initial guess")
plt.ylabel("Number of iterations")
plt.title("Newton Method: Initial Guess vs Number of Iterations")
plt.grid(True)
plt.savefig("initial_guess_vs_iterations.png")
plt.show()


x_values = np.arange(-5, 5, 0.01)
y_values = f(x_values)

plt.figure(figsize=(10, 6))
plt.plot(x_values, y_values, label="f(x) = x sin(4x)")
plt.axhline(0, linewidth=1)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Function f(x) = x sin(4x)")
plt.legend()
plt.grid(True)
plt.savefig("function_graph.png")
plt.show()
```
